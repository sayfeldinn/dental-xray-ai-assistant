# 00 — Setup and source-data check

Run this first. It locates the DENTEX source data (kept outside the repo) and
verifies every file the later notebooks need.

| Task | Source | Images | Boxes |
|---|---|---|---|
| Tooth detector (32 FDI classes) | `quadrant_enumeration` | 634 | 18,095 teeth |
| Disease detector (4 classes) | `quadrant-enumeration-disease` | 705 | 3,529 lesions |
| Locked final judge | official `validation_data` | 50 | 182 lesions |

Nothing here writes to the source data; it is read-only.

```python
import os
import sys
from pathlib import Path

# Repo root is the parent of this notebook directory.
REPO_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(REPO_ROOT / "notebooks"))

import dentex_data as dd

DATA_ROOT = Path(os.environ.get("DENTEX_DATA",
                                "/home/sayfeldin/Projects/dental-model-research"))

PATHS = {
    "QE json":     DATA_ROOT / "data/raw/dentex/quadrant_enumeration/train_quadrant_enumeration.json",
    "QE images":   DATA_ROOT / "data/raw/dentex/quadrant_enumeration/xrays",
    "QED json":    DATA_ROOT / "data/raw/dentex/training_data/quadrant-enumeration-disease/train_quadrant_enumeration_disease.json",
    "QED images":  DATA_ROOT / "data/raw/dentex/training_data/quadrant-enumeration-disease/xrays",
    "val json":    DATA_ROOT / "data/downloads/validation_triple.json",
    "val images":  DATA_ROOT / "data/raw/dentex/validation_data/quadrant_enumeration_disease/xrays",
}

ok = True
for label, path in PATHS.items():
    exists = path.exists()
    ok &= exists
    print(f"{label:12s} {'OK     ' if exists else 'MISSING'} {path}")

assert ok, "Source data missing - set DENTEX_DATA to the dental-model-research root"
print("\nAll source paths present.")
```
